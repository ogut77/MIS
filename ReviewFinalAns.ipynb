# Part 1: Multiple Table Questions

```python
#Q1- List the first name, last name of all students and the grade they received in each enrollment.

txt='''
SELECT Student.StdFirstName, Student.StdLastName, Enrollment.EnrGrade
FROM Student, Enrollment
WHERE Student.StdNo = Enrollment.StdNo
'''
Q1 = psql.sqldf(txt)
Q1
```

```python
#Q2- Show all offering numbers along with the first and last name of the faculty teaching them.

txt='''
SELECT Offering.OfferNo, Faculty.FacFirstName, Faculty.FacLastName
FROM Offering, Faculty
WHERE Offering.FacNo = Faculty.FacNo
'''
Q2 = psql.sqldf(txt)
Q2
```

```python
#Q3- List student first name, last name and the course number they are enrolled in.

txt='''
SELECT Student.StdFirstName, Student.StdLastName, Offering.CourseNo
FROM Student, Enrollment, Offering
WHERE Student.StdNo = Enrollment.StdNo
AND Enrollment.OfferNo = Offering.OfferNo
'''
Q3 = psql.sqldf(txt)
Q3
```

```python
#Q4- Show the course description and the term it is offered in.

txt='''
SELECT Course.CrsDesc, Offering.OffTerm, Offering.OffYear
FROM Course, Offering
WHERE Course.CourseNo = Offering.CourseNo
'''
Q4 = psql.sqldf(txt)
Q4
```

```python
#Q5- List all students enrolled in course IS320 along with their grade.

txt='''
SELECT Student.StdFirstName, Student.StdLastName, Enrollment.EnrGrade
FROM Student, Enrollment, Offering
WHERE Student.StdNo = Enrollment.StdNo
AND Enrollment.OfferNo = Offering.OfferNo
AND Offering.CourseNo = 'IS320'
'''
Q5 = psql.sqldf(txt)
Q5
```

```python
#Q6- Show faculty first name, last name and the course description of what they teach.

txt='''
SELECT Faculty.FacFirstName, Faculty.FacLastName, Course.CrsDesc
FROM Faculty, Offering, Course
WHERE Faculty.FacNo = Offering.FacNo
AND Offering.CourseNo = Course.CourseNo
'''
Q6 = psql.sqldf(txt)
Q6
```

```python
#Q7- List all offerings held in FALL 2016 with the student names enrolled in them.

txt='''
SELECT Offering.OfferNo, Student.StdFirstName, Student.StdLastName
FROM Offering, Enrollment, Student
WHERE Offering.OfferNo = Enrollment.OfferNo
AND Enrollment.StdNo = Student.StdNo
AND Offering.OffTerm = 'FALL'
AND Offering.OffYear = 2016
'''
Q7 = psql.sqldf(txt)
Q7
```

```python
#Q8- Show each course description and how many students are enrolled in it.

txt='''
SELECT Course.CrsDesc, COUNT(Enrollment.StdNo) AS NumStudents
FROM Course, Offering, Enrollment
WHERE Course.CourseNo = Offering.CourseNo
AND Offering.OfferNo = Enrollment.OfferNo
GROUP BY Course.CrsDesc
'''
Q8 = psql.sqldf(txt)
Q8
```

```python
#Q9- List the student name, course number and grade for all SEATTLE students.

txt='''
SELECT Student.StdFirstName, Student.StdLastName, Offering.CourseNo, Enrollment.EnrGrade
FROM Student, Enrollment, Offering
WHERE Student.StdNo = Enrollment.StdNo
AND Enrollment.OfferNo = Offering.OfferNo
AND Student.StdCity = 'SEATTLE'
'''
Q9 = psql.sqldf(txt)
Q9
```

```python
#Q10- Show each faculty member name, their department and the course number they teach.

txt='''
SELECT Faculty.FacFirstName, Faculty.FacLastName, Faculty.FacDept, Offering.CourseNo
FROM Faculty, Offering
WHERE Faculty.FacNo = Offering.FacNo
ORDER BY Faculty.FacLastName
'''
Q10 = psql.sqldf(txt)
Q10
```

# Part 2: Subquery Questions

```python
#Q11- Find all students whose GPA is higher than the average GPA of all students.

txt='''
SELECT StdFirstName, StdLastName, StdGPA
FROM Student
WHERE StdGPA > (SELECT AVG(StdGPA) FROM Student)
'''
Q11 = psql.sqldf(txt)
Q11
```

```python
#Q12- List all faculty members who earn more than the average faculty salary.

txt='''
SELECT FacFirstName, FacLastName, FacSalary
FROM Faculty
WHERE FacSalary > (SELECT AVG(FacSalary) FROM Faculty)
'''
Q12 = psql.sqldf(txt)
Q12
```

```python
#Q13- Find all students who are not enrolled in any course.

txt='''
SELECT StdFirstName, StdLastName
FROM Student
WHERE StdNo NOT IN (SELECT StdNo FROM Enrollment)
'''
Q13 = psql.sqldf(txt)
Q13
```

```python
#Q14- List all courses that have never been offered.

txt='''
SELECT CourseNo, CrsDesc
FROM Course
WHERE CourseNo NOT IN (SELECT CourseNo FROM Offering)
'''
Q14 = psql.sqldf(txt)
Q14
```

```python
#Q15- Find the student with the highest GPA in the IS major.

txt='''
SELECT StdFirstName, StdLastName, StdGPA
FROM Student
WHERE StdGPA = (SELECT MAX(StdGPA) FROM Student WHERE StdMajor = 'IS')
AND StdMajor = 'IS'
'''
Q15 = psql.sqldf(txt)
Q15
```
